```python
#2. Write the a python program that, given an input list of any level of complexity/nestedness, will return the inner most list plus 1. 
#This is to be done with recursion. Note: the input will contain only integers or lists. 

input_list = [1, 2, 3, 4, [5, 6, 7, [8, 9]]]

def recursive_inner_function(current):
    new_list = []
    for item in current:
        if isinstance(item, list):
            return recursive_inner_function(item)
        else:
            new_list.append(item + 1)
    return new_list
    
print(recursive_inner_function(input_list))
```
