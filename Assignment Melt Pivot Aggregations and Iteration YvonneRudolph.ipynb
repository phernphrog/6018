```python
import pandas as pd
import numpy as np
df = pd.read_csv('breast-cancer.data', names=['Class', 'age', 'menopause', 'tumor-size', 'inv-nodes', 'node-caps', 'deg-malig', 'breast', 'breast-quad', 'irradiat'])
df.head()
```

```python
#melt
#very column not listed in columns_to_melt becomes a value in the Type column, 
#and data from these non-listed columns become values in the Values column.
columns_to_melt = ['Class', 'age', 'tumor-size', 'inv-nodes', 'deg-malig']
melted_df = df.melt(id_vars=columns_to_melt, var_name='Type', value_name = 'Values')
melted_df
```

```python
#pivot
#pivot requires a column with unique IDs, so first create a column with unique IDs
df['Unique_ID'] = np.arange(len(df))
df
```

```python
#pivot continued
# Re-melt including the Unique_ID as an id_var
columns_to_melt = ['Unique_ID', 'Class', 'age', 'tumor-size', 'inv-nodes', 'deg-malig']
melted_df = df.melt(id_vars=columns_to_melt, var_name='Type', value_name='Values')

# Pivot using the columns present in melted_df
pivoted = melted_df.pivot(index='Unique_ID', columns='Type', values='Values')
pivoted
```

```python
#aggregation
#since this dataset only contains 1 column with integers, 
#this column (deg-malig) is grouped and aggregated data is listed for it.
aggregate_df = df.groupby('tumor-size').agg({'deg-malig':['mean', 'min', 'max']})
aggregate_df
```

```python
#Iterations
#for each value of 1 in deg-malig, enter 'low-risk' into corresponding risk column, 
#for each value of 2 in deg'malig, enter 'med-risk' into corresponding risk column, 
#and for each value of 3 in deg-malig, enter 'high-risk' into corresponding risk column.
risk = []
for index, row in df.iterrows():
    if row['deg-malig'] == 1:
        risk.append('low-risk')
    elif row['deg-malig'] == 2:
        risk.append('med-risk')
    else:
        row['deg-malig'] == 3
        risk.append('high-risk')

# Add the list as a new column in df
df['risk'] = risk
df[['tumor-size', 'deg-malig', 'risk']].head()
```

```python
#groupby
#groups datasets by the Class column,
#selects only the menopause column,
#and counts the non-null data points in Class for each unique value in class.
grouped = df.groupby('Class')['menopause'].count()
grouped
```
