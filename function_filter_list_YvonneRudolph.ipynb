```python
# 3. Write a Python program that defines a standard function to filter a list of numbers. 
# The function should accept two arguments:
#1 A list of numbers
#2 A user-defined threshold value

num_list = [12, 18, 4, 13, 6, 72, 51, 3, 49, 32, 76, 28, 9, 98]

def short_list(num_list, threshold=20):
    new_list = []
    for num in num_list:
        if num <= threshold:
            new_list.append(num)
    return new_list

x = short_list(num_list)
print(x)
```
