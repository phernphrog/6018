```python
#1. Write a python program that, given an input list of any level of complexity/nestedness, will return the inner most list plus 1. 
#This is to be done with a while loop. Note: the input will contain only integers or lists.

input_list = [1, 2, 3, 4, [5, 6, 7, [8, 9]]]

def while_func(current):
    new_list = []
    i = 0
    while i < len(current):
        if isinstance(current[i], list):
            level_1 = current[i]
            j = 0
            while j < len(level_1):
                if isinstance(level_1[j], list):
                    level_2 = level_1[j]
                    k = 0
                    while k < len(level_2):
                        val = level_2[k] + 1
                        new_list.append(val)
                        k += 1
                j += 1
        i += 1
    return new_list

print(while_func(input_list))
```
