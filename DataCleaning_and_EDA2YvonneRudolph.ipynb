```python
import pandas as pd
import numpy as np
df = pd.read_csv("train.csv")
df_shape = df.shape
print("Shape of data structure:\n", df_shape, "\n")
df.info()
df.head()
display(df.head()), "\n"
df_missing_data = df.isna().sum()
print("Missing Data:\n", df_missing_data, "\n")
df_missing_percent = df.isna().sum() / len(df) * 100
print("Percent of Missing Data per column:\n", df_missing_percent, "\n")
df_duplicates = df.duplicated().sum()
print("Number of duplicate entries:", df_duplicates, "\n")

Q1_age = df["Age"].quantile(0.25)
# print("25% of passengers were this age or or younger:", Q1)
Q3_age = df["Age"].quantile(0.75)
# print("75% of passengers were this age or or younger:", Q3)
Q1_fare = df["Fare"].quantile(0.25)
# print("25% of passengers paid this amount or less:", Q1)
Q3_fare = df["Fare"].quantile(0.75)
# print("75% of passengers paid this amount or less:", Q3)
IQR_age = Q3_age - Q1_age
lower_bound_age = Q1_age - 1.5 * IQR_age
upper_bound_age = Q3_age + 1.5 * IQR_age
IQR_fare = Q3_fare - Q1_fare
lower_bound_fare = Q1_fare - 1.5 * IQR_fare
upper_bound_fare = Q3_fare + 1.5 * IQR_fare
outliers_fare = df[(df['Fare'] < lower_bound_fare) | (df['Fare'] > upper_bound_fare)]
print("Fare Outliers:\n", outliers_fare, "\n")
# There were 116 fare outliers in the group.
outliers_age = df[(df['Age'] < lower_bound_age) | (df['Age'] > upper_bound_age)]
print("Age Outliers:\n", outliers_age)
#There were 11 age outliers in the group, and they were all men.
```

```python
# 2. Clean and Validate the Dataset (25 points)
# Create a copy of the original DataFrame before cleaning.
# Select an appropriate strategy for handling missing values in Age, Cabin, and Embarked (retain, remove, or impute).
# Investigate potential outliers and decide whether to retain, correct, or remove them.
# Check and correct inappropriate data types, if necessary.
# Validate the cleaned dataset by checking missing values, duplicates, and data types again.
# Document each cleaning decision and explain why you selected that approach.
```

```python
# Create a copy of the original DataFrame before cleaning.
df_copy = df.copy()
#Strategy for handling missing data: 
    #"Age"; since there are only 11 age outliers (data tends towards average), replace missing data with mean age.
df_copy['Age'] = df_copy['Age'].fillna(df_copy['Age'].mean())
    #"Cabin"; most data is missing. Since this column is not required for any other part of the assignment, the column will be deleted.
df_copy = df_copy.drop(columns=['Cabin'])
    #"Embarked"; there are only 2 missing values for embarked, removing the rows with missing data will have little impact.
df_copy = df_copy.dropna(subset=['Embarked'])
# Investigate potential outliers in SibSp and Parch and decide whether to retain, correct, or remove them. 
#All data were retained, as none seemed impossible.
Q1_sib = df_copy['SibSp'].quantile(0.25)
Q3_sib = df_copy['SibSp'].quantile(0.75)
IQR_sib = Q3_sib - Q1_sib

lower_bound_sib = Q1_sib - 1.5 * IQR_sib
upper_bound_sib = Q3_sib + 1.5 * IQR_sib

outliers_sibsp = df_copy[(df_copy['SibSp'] < lower_bound_sib) | (df_copy['SibSp'] > upper_bound_sib)]
print("Sibling/spouse outliers:\n", outliers_sibsp)

Q1_parch = df_copy['Parch'].quantile(0.25)
Q3_parch = df_copy['Parch'].quantile(0.75)
IQR_parch = Q3_parch - Q1_parch

lower_bound_parch = Q1_parch - 1.5 * IQR_parch
upper_bound_parch = Q3_parch + 1.5 * IQR_parch

outliers_parch = df_copy[(df_copy['Parch'] < lower_bound_sib) | (df_copy['Parch'] > upper_bound_sib)]
print("Parent/child outliers:\n", outliers_parch)

print(f"Total Sibling/spouse outliers: {len(outliers_sibsp)}")
print(f"Total Parent/child outliers: {len(outliers_parch)}")

# Show 20 highest values, sorted in descending order, to see if outliers appear invalid.
for col in df_copy[['Age', 'SibSp', 'Parch', 'Fare']]:
    print(f"\nColumn: {col}")
    print(f"Top 20 highest values:")
    print(df_copy[col].nlargest(20))  
    
#Check and correct inappropriate data types, if necessary, but there are no inappropriate data types to correct.
for col in df_copy.columns:
    print(f"\n{col}:")
    print(f"  Data type: {df_copy[col].dtype}")
    print(f"  Unique values: {df_copy[col].nunique()}")
    print(f"  Sample values: {df_copy[col].unique()[:5]}")
    print(f"  Null count: {df_copy[col].isnull().sum()}")
#Validate the cleaned dataset by checking missing values, duplicates, and data types again (see above for missing data).
df_copy.info()
df_copy.head()
display(df_copy.head()), "\n"
df_copy_duplicates = df_copy.duplicated().sum()
print("Number of duplicate entries:", df_copy_duplicates, "\n")
```
