```python
#3 Import a dataset with numbers and texts keeping the text intact in python numpy.
import pandas as pd
import requests
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

iris = requests.get(url)
#print(iris)
column_names = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width",
    "species"
]
iris_df = pd.read_csv(url, header=None, names=column_names)
# Display the first five rows
print(iris_df.head())
```

```python
#4 Find the position of the first occurrence of a value greater than 1.0 in petalwidth 4th column of iris dataset.
is_one = iris_df['petal_width'] == 1.0
first_index = np.argmax(is_one)
print(f"First index where petal_width is 1.0: {first_index}")
```
