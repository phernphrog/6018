```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

flights_data = pd.read_csv('flights.csv')
flights_data.head(10)
weather_data_pd = pd.read_csv('weather.csv')
#weather_data_np = weather_data_pd.to_numpy()
#%% Pandas Data Filtering/Sorting Question Answering 
#use flights_data

#flights_data.head()
```

```python
#Question 1 How many flights were there from JFK to SLC? Int
count = len(flights_data.loc[(flights_data['origin'] == 'JFK') & (flights_data['dest'] == 'SLC')])
count
```

```python
#Question 2 How many airlines fly to SLC? Should be int
airlines_to_slc = flights_data.loc[flights_data['dest'] == 'SLC', 'carrier'].nunique()
airlines_to_slc
```

```python
#Question 3 What is the average arrival delay for flights to RDU? float
avg_delay = flights_data.loc[flights_data['dest'] == 'RDU', 'arr_delay'].mean()
avg_delay
```

```python
#Question 4 What proportion of flights to SEA come from the two NYC airports (LGA and JFK)?  float
flights_sea = flights_data[(flights_data['dest'] == 'SEA') & (flights_data['origin'].isin(['LGA', 'JFK']))]
total_to_sea = flights_data[flights_data['dest'] == 'SEA']
proportion = len(flights_sea) / len(total_to_sea)
proportion
```

```python
#Question 5 Which date has the largest average departure delay? Pd slice with date and float
#please make date a column. Preferred format is 2013/1/1 (y/m/d)
flights_data['date'] = pd.to_datetime(flights_data[['year', 'month', 'day']]).dt.strftime('%Y/%m/%d')
avg_delay = flights_data.groupby('date')[['dep_delay']].mean().sort_values(by='dep_delay', ascending=False)
largest_delay = avg_delay[:1]
largest_delay
```

```python
#Question 6 Which date has the largest average arrival delay? pd slice with date and float
avg_delay = flights_data.groupby('date')[['arr_delay']].mean().sort_values(by='arr_delay', ascending=False)
largest_delay = avg_delay[:1]
largest_delay
```

```python
#Question 7 Which flight departing LGA or JFK in 2013 flew the fastest? pd slice with tailnumber and speed
#speed = distance/airtime
lga_jfk_2013 = flights_data[(flights_data['origin'].isin(['LGA', 'JFK'])) & (flights_data['year'] == 2013)].copy()
lga_jfk_2013['speed'] = lga_jfk_2013['distance'] / lga_jfk_2013['air_time']
fastest_plane = lga_jfk_2013.sort_values(by='speed')[:1]
fastest_plane[['tailnum', 'speed']]
```
